```python
from statsbombpy import sb

competitions = sb.competitions()

world_cup = competitions[
    (competitions["competition_name"] == "FIFA World Cup") &
    (competitions["season_name"] == "2022")
]

display(world_cup)
```

```python
competition_id = int(world_cup.iloc[0]["competition_id"])
season_id = int(world_cup.iloc[0]["season_id"])

matches = sb.matches(
    competition_id=competition_id,
    season_id=season_id
)

display(matches.head())
```

```python
competition_id = int(world_cup.iloc[0]["competition_id"])
season_id = int(world_cup.iloc[0]["season_id"])

matches = sb.matches(
    competition_id=competition_id,
    season_id=season_id
)

display(matches.head())
```

```python
match_id = int(matches.iloc[0]["match_id"])

events = sb.events(match_id=match_id)

print(events.columns.tolist())
display(events.head())
```

```python
passes = events[events["type"] == "Pass"].copy()

# In StatsBomb event data, a null pass_outcome
# generally indicates a completed pass.
passes["completed"] = passes["pass_outcome"].isna()

pass_stats = (
    passes.groupby("team")
    .agg(
        total_passes=("completed", "size"),
        completed_passes=("completed", "sum")
    )
)

pass_stats["accuracy_pct"] = (
    pass_stats["completed_passes"]
    / pass_stats["total_passes"] * 100
)

display(pass_stats.round(2))
```
